# Linux 基础命令
## pwd  
查看当前所在目录
## ls / 
可以看到当前目录有什么。默认 `ls` 带颜色是 `--color=auto`，由 `LS_COLORS` 环境变量控制文件类型颜色。其中tmp是临时目录，系统存放临时文件的目录等等。
## ls -l 
查看详细信息
## ls -a 
显示隐藏文件
## cd 
切换目录
## mkdir
创建目录
## touch
创建空文件
## cat
查看文件内容
## echo "内容" > 文件名
向文件写入内容
