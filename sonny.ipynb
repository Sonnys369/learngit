## 3. 基本命令

### chown、chmod
更改文件所有者（owner），也可以同时更改文件所属组。更改文件属性，owner/group/others(拥有者/组/其他) 三种身份各有自己的 read/write/execute 权限。
### pwd
作用：查看当前所在目录。
### ls
查看当前目录内容，ls：普通查看，ls -l：查看详细信息，ls -a：包括隐藏文件。
### mv、rm
重命名以及移动文件;删除文件,Linux 命令行删除通常不会进入回收站。
### 其他
mkdir 创建目录，rmdir删除空目录，cd 切换目录，touch 创建文件，> 覆盖写入，>> 追加写入，cat 由第一行开始显示文件内容(tac从最后一行开始显示)，cp 复制文件
## Linux 路径
绝对路径：从 `/` 开始。
例如：/root/linux-study/hello.txt。
相对路径：相对于当前目录。
特殊路径：
. 当前目录
.. 上一级目录
~ 当前用户的家目录



## 文件查看与搜索

### head
查看文件开头：head -n 3 test.txt
### tail
查看文件末尾：tail -n 3 test.txt ; 或者也可以持续查看文件更新：tail -f file.log。
### less
分页查看文件：/关键词：搜索，q：退出，感觉在内容量比较大的情况下整体查看更方便。
### grep
搜索文件内容：grep linux test.txt；加上-n可以显示出来具体的行号，加上-i可以忽略大小写。
### find
主要用来查找文件
### 管道
| 将前一个命令的输出作为后一个命令的输入。ls /etc | grep ssh可以理解为先把/etc列出来然后在结果中寻找包含ssh的内容。

```python
mkdir linux-study
cd linux-study
touch hello.txt
echo "hello linux" > hello.txt
echo "I am learning Linux" >> hello.txt
cat hello.txt
```

```python
cp hello.txt hello-copy.txt
mkdir backup
cp hello.txt backup/
mv hello-copy.txt note.txt
mv note.txt backup/
ls
ls backup
cd ..
pwd
cd ~/linux-study
pwd
```

### 注意问题：
1.直接执行 cat，终端会等待标准输入。原因：没有指定文件时，cat 会读取标准输入并原样输出。可用 Ctrl + D 结束输入。
2.主要遇到了 SSH 本地连接异常的问题。已排查：sshd 服务状态，Linux 防火墙，腾讯云防火墙，SSH 监听端口。
