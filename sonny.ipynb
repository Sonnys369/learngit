# SSH 公钥认证

## 1. SSH 密钥

SSH 公钥认证使用密钥对：

* 私钥：保存在客户端，必须保密。
* 公钥：可以公开，通常放在服务器用户的 `~/.ssh/authorized_keys` 中。

典型结构：

```text
客户端
├── 私钥
└── ssh
       ↓
服务器
└── ~/.ssh/authorized_keys
        └── 公钥
```

## 2. SSH 用户密钥和服务器 Host Key

`authorized_keys`：

> 服务器用于验证客户端用户身份。

`known_hosts`：

> 客户端用于识别服务器身份。

两者用途不同。

## 3. 生成测试密钥

```bash
ssh-keygen -t ed25519 -f /tmp/linuxlab-demo -N '' -C 'linuxlab-demo'
```

生成：

```text
/tmp/linuxlab-demo
/tmp/linuxlab-demo.pub
```

其中前者是私钥，后者是公钥。

## 4. 配置 authorized_keys

```bash
install -d -m 700 -o linuxlab -g linuxlab /home/linuxlab/.ssh

install -m 600 -o linuxlab -g linuxlab \
/tmp/linuxlab-demo.pub \
/home/linuxlab/.ssh/authorized_keys
```

## 5. 测试公钥登录

```bash
ssh \
-o PasswordAuthentication=no \
-o KbdInteractiveAuthentication=no \
-o UserKnownHostsFile=/tmp/linuxlab-known_hosts \
-i /tmp/linuxlab-demo \
linuxlab@127.0.0.1
```

如果成功，可以不输入用户密码直接登录。

## 6. SSH 配置排查

查看配置文件：

```bash
grep -E '^(Port|PubkeyAuthentication|PasswordAuthentication|PermitRootLogin)' /etc/ssh/sshd_config
```

查看最终生效配置：

```bash
sshd -T
```

检查 SSH 配置语法：

```bash
sshd -t
```

## 7. SSH 排障思路

```text
网络
 ↓
端口
 ↓
sshd
 ↓
认证方式
 ↓
私钥
 ↓
authorized_keys
 ↓
文件权限
 ↓
用户权限
```

## 8. 安全注意事项

* 私钥不能上传到服务器或提交 GitHub。
* 不要删除正在使用的 `/root/.ssh/authorized_keys`。
* 修改 SSH 配置后应先使用 `sshd -t` 检查语法。
* 远程服务器上不要随意停止 `sshd`。
* root 用户执行 `rm -rf`、`chmod -R`、`chown -R` 时需要特别谨慎。
